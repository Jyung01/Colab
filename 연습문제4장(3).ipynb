```python
n = 0
for score in [60, 90, 85, 99, 75]:
  n += 1
  if score >= 90:
    print("%d번 학생은 A 입니다." %n)
  elif score >= 80 and score < 90:
    print("%d번 학생은 B 입니다." %n)
  else:
    print("%d번 학생은 C 입니다." %n)
```

```python
def pow1(*args):
  x = 0
  for i in args:
    x += i * i
  return x

pow1(4, 2)
```

```python
import matplotlib.pyplot as plt

x = [2016, 2017, 2018, 2019, 2020]
y = [350, 410, 520, 695, 543]
plt.plot(x, y)
plt.title('Annual sales')
plt.xlabel('years')
plt.ylabel('sales')
plt.show()
```

```python
y1 = [350, 410, 520, 695]
y2 = [200, 250, 385, 350]
x = range(len(y1))
plt.bar(x, y1, width = 0.7, color = "blue")
plt.bar(x, y2, width = 0.7, color = "red", bottom = y1)
plt.title('Quarterly sales')
plt.xlabel('Quarters')
plt.ylabel('sales')
xLabel = ['first', 'second', 'third', 'fourth']
plt.xticks(x, xLabel, fontsize = 10)
plt.legend(['chairs', 'desks'])
plt.show()
```
