```python
import pandas as pd

wine_header = ['Class', 'Alcohol', 'Malic acid', 'Ash', 'Alcalinity of ash', 'Magnesium'
              ,'Total phenols', 'Flavanoids', 'Nonflavanoid phenols', 'Proanthocyanins'
              , 'Color intensity', 'Hue', 'OD280/OD315 of diluted wines', 'Proline']
wine_df = pd.read_csv('winedata.txt', sep = ',', names = wine_header, header = 0, engine = 'python')
wine_df.to_csv('winedata.csv', index = False)
```

```python
wine_df.columns = wine_df.columns.str.replace(' ', '_')
```
