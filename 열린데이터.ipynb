```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# API 주소 및 필요한 매개변수 설정
url = 'http://openAPI.seoul.go.kr:8088/{0}/xml/ChunmanFreeSuggestions/1/5'
api_key = ''  # 발급받은 API Key 입력

# API 호출
response = requests.get(url.format(api_key))

# XML 데이터 파싱
soup = BeautifulSoup(response.text, 'lxml')
rows = soup.find_all('row')

# DataFrame 생성
data = []
for row in rows:
    item = {}
    for tag in row.find_all():
        item[tag.name] = tag.text
    data.append(item)
df = pd.DataFrame(data)
print(df)

df.to_csv('suggestions_data.csv', index=False, encoding = 'cp949')
```
