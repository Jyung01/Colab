```python
from bs4 import BeautifulSoup
import urllib.request
import pandas as pd

result = []

for page in range(1, 10):
    url = 'https://pelicana.co.kr/store/stroe_search.html?page=%d&branch_name=&gu=&si=' %page
    html = urllib.request.urlopen(url)
    soup = BeautifulSoup(html, 'html.parser')
    tag_table = soup.find('table', {'class': 'table mt20'})
    tag_tbody = tag_table.find('tbody')
    for store in tag_tbody.find_all('tr'):
        store_td = store.find_all('td')
        store_name = store_td[0].string
        store_address = store_td[1].string
        store_phone = store_td[2].string
        result.append([store_name, store_address, store_phone])

pelicana_tbl = pd.DataFrame(result, columns=('store', 'address', 'phone'))
pelicana_tbl.to_csv('pelicana.csv', encoding='utf-8-sig', mode='w', index=False)
```
