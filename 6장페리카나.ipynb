```python
from bs4 import BeautifulSoup
import urllib.request
import pandas as pd

result = []

for page in range(1, 107):
  url = 'https://pelicana.co.kr/store/stroe_search.html?page=%d&branch_name=&gu=&si=' %page
  print(url)
  html = urllib.request.urlopen(url)
  soupPeli = BeautifulSoup(html, 'html.parser')
  tag_tbody = soupPeli.find('tbody')
  for store in tag_tbody.find_all('tr'):
    if len(store) <= 2:
      break
    store_td = store.find_all('td')
    store_name = store_td[0].string
    store_address = store_td[1].string
    store_phone = store_td[2].string
    result.append([store_name]+[store_address]+[store_phone])

print(len(result))

result[0]

peli_tbl = pd.DataFrame(result, columns = ('매장이름', '주소', '전화번호'))
print(peli_tbl)
peli_tbl.to_csv("peli.csv", encoding = "cp949", mode = "w", index = True)
```
