```python
from bs4 import BeautifulSoup
import urllib.request
import ssl

result = []

url = 'https://sbiz.wooribank.com/biz/Dream?withyou=BZFXD0020'

# SSL 보안 수준 설정 변경
ssl_context = ssl.create_default_context()
ssl_context.set_ciphers("DEFAULT@SECLEVEL=1")

# SSL 컨텍스트를 사용하여 URL 열기
html = urllib.request.urlopen(url, context=ssl_context)
soup_ex = BeautifulSoup(html, 'html.parser')
print(soup_ex)
tag_find = soup_ex.find('table', class_='tbl-type-1 txt-c mb10 ui-set-tbl-type')
print(tag_find)
```
